```python
# ============================== Assignment 9 ==============================
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import statsmodels.api as sm
from scipy import stats
from scipy.cluster.hierarchy import linkage, dendrogram, fcluster
from scipy.spatial.distance import pdist, squareform
from sklearn.cluster import KMeans
from sklearn.decomposition import PCA
from statsmodels.graphics.tsaplots import plot_acf, plot_pacf
from statsmodels.tsa.seasonal import seasonal_decompose
from statsmodels.tsa.arima.model import ARIMA

mtcars = sm.datasets.get_rdataset("mtcars", "datasets").data

# Q1
selected_vars = mtcars[["mpg", "disp", "hp", "drat", "wt", "qsec"]]
print(selected_vars.describe())

# Q2
scaled_vars = (selected_vars - selected_vars.mean()) / selected_vars.std()
print(scaled_vars)
print("Standardization is needed because the variables are on very different scales (e.g. disp vs drat), so unscaled PCA would be dominated by the variables with the largest variance.")

# Q3
cars_pca = PCA().fit(scaled_vars)
pca_scores = cars_pca.transform(scaled_vars)
print(pd.DataFrame(pca_scores, index=scaled_vars.index, columns=[f"PC{i}" for i in range(1, 7)]).head())
print("Standard deviations:", np.sqrt(cars_pca.explained_variance_))

# Q4
eigenvalues = cars_pca.explained_variance_
prop_variance = eigenvalues / eigenvalues.sum()
cum_variance = np.cumsum(prop_variance)
print(pd.DataFrame({"Eigenvalue": eigenvalues, "Proportion": prop_variance, "Cumulative": cum_variance}, index=[f"PC{i}" for i in range(1, 7)]))

# Q5
pca_summary = pd.DataFrame({
    "Standard deviation": np.sqrt(eigenvalues),
    "Proportion of Variance": prop_variance,
    "Cumulative Proportion": cum_variance
}, index=[f"PC{i}" for i in range(1, 7)]).T
print(pca_summary.round(4))
print("Minimum PCs for at least 80% variance:", int(np.argmax(cum_variance >= 0.80) + 1))

# Q6
plt.figure(figsize=(7, 4))
plt.plot(range(1, 7), eigenvalues, "o-")
plt.title("Scree Plot")
plt.xlabel("Principal Component")
plt.ylabel("Variance (Eigenvalue)")
plt.xticks(range(1, 7))
plt.show()

# Q7
loadings = pd.DataFrame(cars_pca.components_.T, index=scaled_vars.columns, columns=[f"PC{i}" for i in range(1, 7)])
print(loadings.round(4))
print("Strongest contributors to PC1:", loadings["PC1"].abs().sort_values(ascending=False).index[:3].tolist())
print("Strongest contributors to PC2:", loadings["PC2"].abs().sort_values(ascending=False).index[:3].tolist())

# Q8
plt.figure(figsize=(8, 7))
plt.scatter(pca_scores[:, 0], pca_scores[:, 1], alpha=0)
for name, (a, b) in zip(scaled_vars.index, pca_scores[:, :2]):
    plt.text(a, b, name, fontsize=8, color="grey", ha="center")
scale_factor = np.abs(pca_scores[:, :2]).max()
for var, (a, b) in zip(scaled_vars.columns, loadings[["PC1", "PC2"]].values):
    plt.arrow(0, 0, a * scale_factor * 0.9, b * scale_factor * 0.9, color="red", head_width=0.08, length_includes_head=True)
    plt.text(a * scale_factor, b * scale_factor, var, color="red", fontsize=11, ha="center")
plt.axhline(0, color="black", linewidth=0.5)
plt.axvline(0, color="black", linewidth=0.5)
plt.xlabel("PC1")
plt.ylabel("PC2")
plt.title("PCA Biplot for mtcars Variables")
plt.show()
```
